```python
# Import pandas for working with our time-series dataset.
import pandas as pd

# Load the weekly demand data created during data preparation.
df = pd.read_parquet("../artifacts/weekly.parquet")

# Make sure the data is sorted chronologically within each product.
df = df.sort_values(["item", "date"]).reset_index(drop=True)

# Display the earliest and latest dates in the dataset.
print("Start date:", df["date"].min())
print("End date:", df["date"].max())
```

```python
# Find the final date in our dataset.
max_date = df["date"].max()

# The final 26 weeks will be our untouched test period.
test_start = max_date - pd.Timedelta(weeks=26)

# Everything before the test period is used for training/validation.
train_df = df[df["date"] < test_start].copy()

# The final 26 weeks are kept completely separate for final evaluation.
test_df = df[df["date"] >= test_start].copy()

# Print the resulting date ranges and sizes.
print("Training period:")
print(train_df["date"].min(), "to", train_df["date"].max())
print("Rows:", len(train_df))

print("\nTest period:")
print(test_df["date"].min(), "to", test_df["date"].max())
print("Rows:", len(test_df))
```
